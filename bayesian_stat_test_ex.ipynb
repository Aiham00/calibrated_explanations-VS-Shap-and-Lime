```python
def bayesian_test_ex(df,ex1,ex2, dataset = None, model = None, mu1 =0.98, sigma1 = 0.2, mu2 =0.98, sigma2 = 0.2, val = "stab"):
    
    if dataset is not None:
        df = df[df["dataset"]==dataset]
    model1_values = df[ex1]
    model2_values = df[ex2]

    with pm.Model() as model:
        
        # Priors for the means and standard deviations
        mu_model1 = pm.Normal('mu_model1', mu=mu1, sigma=sigma1)  # Prior for value mean in model1
        mu_model2 = pm.Normal('mu_model2', mu=mu2, sigma=sigma2)  # Prior for value mean in model2
        sigma_model1 = pm.HalfNormal('sigma_model1', sigma=sigma1)  # Prior for standard deviation in  model1
        sigma_model2 = pm.HalfNormal('sigma_model2', sigma=sigma2)  # Prior for standard deviation in  model2

        # Likelihood (data likelihood assuming normal distribution)
        value_model1_obs = pm.Normal('value_model1_obs', mu=mu_model1, sigma=sigma_model1, observed=model1_values)
        value_model2_obs = pm.Normal('value_model2_obs', mu=mu_model2, sigma=sigma_model2, observed=model2_values)
        
        # Sampling from the posterior distribution
        trace = pm.sample(2000, tune=500, chains=1, cores=1,progressbar=False,
                            compute_convergence_checks=False)# 2000 looks good number but can be changed

    # Compute the probability that the value in model1 is higher than in the model2
    posterior_diff = trace.posterior['mu_model1'] - trace.posterior['mu_model2']
    prob_higher = (posterior_diff > 0).mean().item()
    if val in ["stab", "rob"]:
        val = "variance represinting " + val

    if dataset is not None: 
        print(f"Posterior probability that {val} is higher in {ex1} than in {ex2} for the dataset {dataset}: {prob_higher:.2f}")
    else:
        print(f"Posterior probability that {val} is higher in {ex1} than in {ex2}: {prob_higher:.2f}")

```

```python
ce_ex_s = ["CE",	"CCE"]
all_ex_s = ["CE",	"CCE",	"L C",	"S C",	"L U",	"S U"]
```

```python
df_stab = pd.read_csv(os.path.join("evaluation","stability_all.csv"))
df_stab_pr = df_stab.iloc[0:2]
val_stab = "stab"
metrics_stab = {}
for value in all_ex_s:
    metrics_stab[value]=[df_stab_pr[value].mean(),sqrt(df_stab_pr[value].var())]
metrics_stab
```

```python
df_rob = pd.read_csv(os.path.join("evaluation","rob_all.csv"))
df_rob_pr = df_rob.iloc[0:2]
val_rob = "rob"
metrics_rob = {}
for value in all_ex_s:
    metrics_rob[value]=[df_rob_pr[value].mean(),sqrt(df_rob_pr[value].var())]
metrics_rob
```

```python
df_time = pd.read_csv(os.path.join("evaluation","stability_time.csv"))
df_time_pr = df_time.iloc[0:2]
val_time = "time"
metrics_time = {}
for value in all_ex_s:
    metrics_time[value]=[df_time_pr[value].mean(),sqrt(df_time_pr[value].var())]
metrics_time
```

```python
ex1 = "CE"
ex2 = "L C"
bayesian_test_ex(df_stab,ex1,ex2,mu1=metrics_stab[ex1][0],sigma1=metrics_stab[ex1][1],mu2=metrics_stab[ex2][0],sigma2=metrics_stab[ex2][1],val=val_stab)
bayesian_test_ex(df_rob,ex1,ex2,mu1=metrics_rob[ex1][0],sigma1=metrics_rob[ex1][1],mu2=metrics_rob[ex2][0],sigma2=metrics_rob[ex2][1],val=val_rob)
bayesian_test_ex(df_time,ex1,ex2,mu1=metrics_time[ex1][0],sigma1=metrics_time[ex1][1],mu2=metrics_time[ex2][0],sigma2=metrics_time[ex2][1],val=val_time)
```
